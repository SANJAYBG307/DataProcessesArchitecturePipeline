## Practice Dataset Setup

Run this setup cell first. It creates a reproducible sample dataset used by the techniques below.


```python
# 7.0 Practice Dataset Setup
# Definition: Create a reproducible sample dataset used across all examples in this notebook.
# Example: Build the sample DataFrame and confirm it loads before running section techniques.

# Import pandas
import pandas as pd
# Import numpy
import numpy as np
# Import matplotlib.pyplot
import matplotlib.pyplot as plt
# Import seaborn
import seaborn as sns
# Compute rng
rng = np.random.default_rng(7)
# Compute n
n = 120
# Compute dates
dates = pd.date_range('2025-01-01', periods=n, freq='D')
# Create the sample DataFrame
df = pd.DataFrame({
 'OrderDate': dates,
 'Region': rng.choice(['North', 'South', 'West'], n),
 'Product': rng.choice(['Laptop', 'Mouse', 'Keyboard'], n),
 'Quantity': rng.integers(1, 5, n),
 'UnitPrice': rng.choice([25, 45, 700, 750], n).astype(float)
})
# Run this transformation step
df['Sales'] = df['Quantity'] * df['UnitPrice']
# Run this transformation step
df.loc[[4, 15], 'Sales'] = np.nan
# Evaluate this expression output
print(df.head())
```

## 7.1 Dataset overview

**Concept:** Inspect summary statistics and representative rows.



```python
# 7.1 Dataset Overview
# Definition: Perform dataset overview to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Evaluate this expression output
print(df.head())
# Evaluate this expression output
print(df.describe(include='all'))
```

## 7.2 Row & column analysis

**Concept:** Check dimensions and variable names.



```python
# 7.2 Row & Column Analysis
# Definition: Inspect column names, null counts, data types, and memory usage.
# Example: List column names and review df.info() output for structure checks.

# Read dataset dimensions (rows, columns)
print(df.shape)
# Extract all column names
print(df.columns.tolist())
```

## 7.3 Data type analysis

**Concept:** Confirm dates, categories and measures have correct types.



```python
# 7.3 Data Type Analysis
# Definition: Verify inferred data types and identify fields requiring conversion.
# Example: Run the cell and interpret the output against the section objective.

# Review inferred data types
print(df.dtypes)
```

## 7.4 Missing value analysis

**Concept:** Quantify missingness and visualize it.



```python
# 7.4 Missing Value Analysis
# Definition: Quantify missing values to prioritize data-quality fixes.
# Example: Run the cell and interpret the output against the section objective.

# Count missing values by column
print(df.isna().sum())
# Run this transformation step
sns.heatmap(df.isna(), cbar=False)
# Run this transformation step
plt.title('Missing values by row and column')
# Run this transformation step
plt.show()
```

## 7.5 Duplicate analysis

**Concept:** Identify repeated full records or business keys.



```python
# 7.5 Duplicate Analysis
# Definition: Detect repeated rows or repeated business keys that can bias analysis.
# Example: Compare full-row duplicates with key-level duplicates.

# Count duplicate rows or duplicate keys
print('Duplicate rows:', df.duplicated().sum())
# Count duplicate rows or duplicate keys
print('Duplicate dates:', df['OrderDate'].duplicated().sum())
```

## 7.6 Distribution analysis

**Concept:** Use histograms and density estimates to examine shapes.



```python
# 7.6 Distribution Analysis
# Definition: Explore distributions, patterns, and relationships before conclusions.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
sns.histplot(data=df, x='Sales', kde=True)
# Run this transformation step
plt.title('Sales distribution')
# Run this transformation step
plt.show()
```

## 7.7 Univariate analysis

**Concept:** Study one variable at a time using summaries and plots.



```python
# 7.7 Univariate Analysis
# Definition: Perform univariate analysis to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Compute frequency distribution
print(df['Region'].value_counts())
# Run this transformation step
sns.countplot(data=df, x='Region')
# Run this transformation step
plt.show()
```

## 7.8 Bivariate analysis

**Concept:** Study associations between two variables.



```python
# 7.8 Bivariate Analysis
# Definition: Perform bivariate analysis to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
sns.scatterplot(data=df, x='Quantity', y='Sales', hue='Region')
# Run this transformation step
plt.show()
```

## 7.9 Outlier analysis

**Concept:** Use box plots and robust summaries to flag unusual values.



```python
# 7.9 Outlier Analysis
# Definition: Perform outlier analysis to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
sns.boxplot(data=df, x='Region', y='Sales')
# Run this transformation step
plt.show()
```

## 7.10 Group / segment analysis

**Concept:** Compare distributions and metrics across business segments.



```python
# 7.10 Group / Segment Analysis
# Definition: Reshape data into analysis-ready structures while preserving correctness.
# Example: Run the cell and interpret the output against the section objective.

# Evaluate this expression output
print(df.groupby('Region', observed=True)['Sales'].agg(['count', 'mean', 'median', 'sum']))
# Run this transformation step
sns.boxplot(data=df, x='Product', y='Sales')
# Run this transformation step
plt.show()
```

## 7.11 Relationship analysis

**Concept:** Inspect correlations and pairwise patterns while considering confounding.



```python
# 7.11 Relationship Analysis
# Definition: Explore distributions, patterns, and relationships before conclusions.
# Example: Run the cell and interpret the output against the section objective.

# Compute numeric
numeric = df[['Quantity', 'UnitPrice', 'Sales']]
# Run this transformation step
sns.heatmap(numeric.corr(), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
# Run this transformation step
plt.show()
```

## 7.12 Time-series analysis

**Concept:** Sort timestamps, set frequency and aggregate over time.



```python
# 7.12 Time-series Analysis
# Definition: Perform time-series analysis to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Compute daily
daily = df.set_index('OrderDate').sort_index()['Sales'].resample('W').sum()
# Run this transformation step
daily.plot(title='Weekly sales')
# Run this transformation step
plt.ylabel('Sales')
# Run this transformation step
plt.show()
```

## 7.13 Trend analysis

**Concept:** Smooth or aggregate time series to reveal longer-term direction.



```python
# 7.13 Trend Analysis
# Definition: Explore distributions, patterns, and relationships before conclusions.
# Example: Run the cell and interpret the output against the section objective.

# Compute daily
daily = df.set_index('OrderDate').sort_index()['Sales'].resample('D').sum()
# Run this transformation step
daily.rolling(7, min_periods=1).mean().plot(title='7-day rolling sales')
# Run this transformation step
plt.show()
```

## 7.14 Seasonality analysis

**Concept:** Compare recurring patterns by month, weekday or other known cycle.



```python
# 7.14 Seasonality Analysis
# Definition: Explore distributions, patterns, and relationships before conclusions.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
df['Weekday'] = df['OrderDate'].dt.day_name()
# Group records by key before summarization
weekday_sales = df.groupby('Weekday', observed=True)['Sales'].mean().reindex(['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'])
# Run this transformation step
weekday_sales.plot(kind='bar', title='Average sales by weekday')
# Run this transformation step
plt.show()
```

## 7.15 Anomaly analysis

**Concept:** Flag unusual values using transparent rules, then investigate context.



```python
# 7.15 Anomaly Analysis
# Definition: Apply a data-mining method and evaluate model behavior.
# Example: Run the cell and interpret the output against the section objective.

# Compute s
s = df['Sales']
# Compute median
median = s.median()
# Compute mad
mad = (s - median).abs().median()
# Compute flag
flag = (s - median).abs() > 3 * mad if mad else pd.Series(False, index=s.index)
# Evaluate this expression output
print(df.loc[flag, ['OrderDate', 'Region', 'Product', 'Sales']])
```
