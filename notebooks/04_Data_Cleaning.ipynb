## Practice Dataset Setup

Run this setup cell first. It creates a reproducible sample dataset used by the techniques below.


```python
# 4.0 Practice Dataset Setup
# Definition: Create a reproducible sample dataset used across all examples in this notebook.
# Example: Build the sample DataFrame and confirm it loads before running section techniques.

# Import pandas
import pandas as pd
# Import numpy
import numpy as np
# Create the sample DataFrame
df = pd.DataFrame({
    'OrderID': [1, 2, 2, 4],
    'Customer Name ': [' Asha ', 'RAVI', 'RAVI', 'Meera!'],
    'OrderDate': ['2026-01-01', '2026/01/02', '2026/01/02', 'not-a-date'],
    'Quantity': ['1', '2', '2', 'three'],
    'Sales': [100.0, np.nan, np.nan, -5.0],
    'UnusedColumn': ['x', 'y', 'y', 'z']
})
# Create a copy and assign it to raw_df
raw_df = df.copy()
```

## 4.1 Remove duplicates

**Concept:** Drop duplicate rows or duplicates by business key after choosing a retention rule.



```python
# 4.1 Remove Duplicates
# Definition: Detect repeated rows or repeated business keys that can bias analysis.
# Example: Compare full-row duplicates with key-level duplicates.

# Compute df
df = df.drop_duplicates()
# Evaluate this expression output
print(df)
```

## 4.2 Handle missing / NULL values

**Concept:** Choose dropping or imputation based on context and missingness.



```python
# 4.2 Handle Missing / NULL Values
# Definition: Quantify missing values to prioritize data-quality fixes.
# Example: Report both null counts and null percentages by column.

# Run this transformation step
df['Sales_filled'] = df['Sales'].fillna(df['Sales'].median())
# Evaluate this expression output
print(df[['Sales', 'Sales_filled']])
```

## 4.3 Correct data types

**Concept:** Convert strings to numeric or dates; coerce invalid values to missing for review.



```python
# 4.3 Correct Data Types
# Definition: Verify inferred data types and identify fields requiring conversion.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
df['Quantity_num'] = pd.to_numeric(df['Quantity'], errors='coerce')
# Parse text values into datetime and mark invalid entries as NaT
df['OrderDate_parsed'] = pd.to_datetime(df['OrderDate'], errors='coerce')
# Evaluate this expression output
print(df[['Quantity', 'Quantity_num', 'OrderDate_parsed']])
```

## 4.4 Standardize text values

**Concept:** Use mappings and string methods to normalize categories.



```python
# 4.4 Standardize Text Values
# Definition: Apply corrective transformations to improve data quality.
# Example: Run the cell and interpret the output against the section objective.

# Convert data type for consistent analysis
df['Customer_clean'] = df['Customer Name '].astype('string').str.strip().str.title()
# Evaluate this expression output
print(df['Customer_clean'])
```

## 4.5 Trim whitespace

**Concept:** Remove leading and trailing spaces.



```python
# 4.5 Trim Whitespace
# Definition: Perform trim whitespace to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Convert data type for consistent analysis
df['Customer_trimmed'] = df['Customer Name '].astype('string').str.strip()
# Evaluate this expression output
print(df['Customer_trimmed'].tolist())
```

## 4.6 Standardize date / time

**Concept:** Parse dates explicitly; choose formats and timezone handling based on source semantics.



```python
# 4.6 Standardize Date / Time
# Definition: Apply corrective transformations to improve data quality.
# Example: Run the cell and interpret the output against the section objective.

# Parse text values into datetime and mark invalid entries as NaT
df['OrderDate_clean'] = pd.to_datetime(df['OrderDate'], errors='coerce')
# Evaluate this expression output
print(df['OrderDate_clean'])
```

## 4.7 Handle invalid values

**Concept:** Flag invalid values first, then correct, replace or quarantine according to rules.



```python
# 4.7 Handle Invalid Values
# Definition: Perform handle invalid values to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
df['Sales_valid'] = df['Sales'].where(df['Sales'].ge(0))
# Evaluate this expression output
print(df[['Sales', 'Sales_valid']])
```

## 4.8 Remove unwanted columns

**Concept:** Drop irrelevant or sensitive columns only when permitted and documented.



```python
# 4.8 Remove Unwanted Columns
# Definition: Inspect column names, null counts, data types, and memory usage.
# Example: List column names and review df.info() output for structure checks.

# Compute clean
clean = df.drop(columns=['UnusedColumn'])
# Evaluate this expression output
print(clean.columns.tolist())
```

## 4.9 Rename columns

**Concept:** Use consistent, clear column names.



```python
# 4.9 Rename Columns
# Definition: Inspect column names, null counts, data types, and memory usage.
# Example: List column names and review df.info() output for structure checks.

# Compute df
df = df.rename(columns={'Customer Name ': 'CustomerName'})
# Extract all column names
print(df.columns.tolist())
```

## 4.10 Remove special / unwanted characters

**Concept:** Use string replacement or regular expressions, preserving legitimate symbols.



```python
# 4.10 Remove Special / Unwanted Characters
# Definition: Perform remove special / unwanted characters to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Convert data type for consistent analysis
df['Customer_alnum'] = df['Customer Name '].astype('string').str.replace(r'[^A-Za-z ]', '', regex=True)
# Evaluate this expression output
print(df['Customer_alnum'])
```

## 4.11 Standardize case

**Concept:** Use consistent casing for text fields.



```python
# 4.11 Standardize Case
# Definition: Apply corrective transformations to improve data quality.
# Example: Run the cell and interpret the output against the section objective.

# Convert data type for consistent analysis
df['Customer_upper'] = df['Customer Name '].astype('string').str.strip().str.upper()
# Evaluate this expression output
print(df['Customer_upper'])
```
