## Practice Dataset Setup

Run this setup cell first. It creates a reproducible sample dataset used by the techniques below.


```python
# 3.0 Practice Dataset Setup
# Definition: Create a reproducible sample dataset used across all examples in this notebook.
# Example: Build the sample DataFrame and confirm it loads before running section techniques.

# Import pandas
import pandas as pd
# Import numpy
import numpy as np
# Create the sample DataFrame
df = pd.DataFrame({
    'OrderID': [1, 2, 2, 4],
    'Customer': ['Asha', None, 'Ravi', 'Meera'],
    'Quantity': [1, 0, -2, 3],
    'Sales': [100.0, 50.0, -20.0, np.nan],
    'OrderDate': ['2026-01-01', '2026-01-02', 'bad-date', '2020-01-01'],
    'CustomerID': [10, 11, 99, 12]
})
# Create the sample DataFrame
customers = pd.DataFrame({'CustomerID': [10, 11, 12]})
```

## 3.1 Completeness

**Concept:** Measure required fields that are populated.



```python
# 3.1 Completeness
# Definition: Assess whether required fields are populated for each record.
# Example: Run the cell and interpret the output against the section objective.

# Compute required
required = ['OrderID', 'Customer', 'Sales']
# Compute mean to assess central tendency
print(df[required].notna().mean().mul(100).round(1))
```

## 3.2 Validity

**Concept:** Check values against business rules and allowed ranges.



```python
# 3.2 Validity
# Definition: Verify values against business rules and allowed ranges.
# Example: Run the cell and interpret the output against the section objective.

# Compute valid_quantity
valid_quantity = df['Quantity'].gt(0)
# Compute valid_sales
valid_sales = df['Sales'].ge(0)
# Evaluate this expression output
print('Invalid quantity rows:', (~valid_quantity).sum())
# Evaluate this expression output
print('Invalid sales rows:', (~valid_sales & df['Sales'].notna()).sum())
```

## 3.3 Consistency

**Concept:** Check cross-field and standardized-value rules.



```python
# 3.3 Consistency
# Definition: Check logical relationships between related fields.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
df['ExpectedSales'] = df['Quantity'] * 100
# Evaluate this expression output
print(df[['Quantity', 'Sales', 'ExpectedSales']])
# Evaluate this expression output
print('Mismatches:', (df['Sales'].notna() & ~np.isclose(df['Sales'], df['ExpectedSales'], equal_nan=True)).sum())
```

## 3.4 Uniqueness

**Concept:** Validate primary or business keys.



```python
# 3.4 Uniqueness
# Definition: Confirm identifiers are unique where uniqueness is expected.
# Example: Run the cell and interpret the output against the section objective.

# Evaluate this expression output
print('Duplicate OrderID rows:', df['OrderID'].duplicated(keep=False).sum())
```

## 3.5 Timeliness / freshness

**Concept:** Parse timestamps and compare against an agreed freshness threshold; source metadata may be needed.



```python
# 3.5 Timeliness / Freshness
# Definition: Evaluate whether records are recent enough for the use case.
# Example: Run the cell and interpret the output against the section objective.

# Parse text values into datetime and mark invalid entries as NaT
dates = pd.to_datetime(df['OrderDate'], errors='coerce')
# Compute reference_date
reference_date = pd.Timestamp('2026-01-10')
# Compute age_days
age_days = (reference_date - dates).dt.days
# Evaluate this expression output
print(age_days)
# Evaluate this expression output
print('Older than 30 days:', (age_days > 30).sum())
```

## 3.6 Data type validity

**Concept:** Attempt safe conversions and count failures.



```python
# 3.6 Data Type Validity
# Definition: Verify inferred data types and identify fields requiring conversion.
# Example: Run the cell and interpret the output against the section objective.

# Compute quantity_num
quantity_num = pd.to_numeric(df['Quantity'], errors='coerce')
# Parse text values into datetime and mark invalid entries as NaT
date_parsed = pd.to_datetime(df['OrderDate'], errors='coerce')
# Count missing values by column
print('Invalid quantities:', quantity_num.isna().sum())
# Count missing values by column
print('Invalid dates:', date_parsed.isna().sum())
```

## 3.7 Referential integrity

**Concept:** Ensure foreign keys exist in the parent/reference table.



```python
# 3.7 Referential Integrity
# Definition: Ensure foreign-key values match available master/reference records.
# Example: Run the cell and interpret the output against the section objective.

# Compute orphans
orphans = df.loc[~df['CustomerID'].isin(customers['CustomerID'])]
# Evaluate this expression output
print(orphans[['OrderID', 'CustomerID']])
```
