# 2) Data Profiling

This notebook is structured as one profiling technique per section.
Each technique includes: concept, runnable example, and a quick hands-on prompt.

Profiling describes the current data state before cleaning or transformation.


## Practice Dataset Setup

Run this setup cell first. It creates a reproducible sample dataset used by the profiling techniques below.


```python
# 2.0 Practice Dataset Setup
# Definition: Create a reproducible sample dataset used across all examples in this notebook.
# Example: Build the sample DataFrame and confirm it loads before running section techniques.

# Import pandas
import pandas as pd
# Import numpy
import numpy as np
# Create the sample DataFrame
df = pd.DataFrame({
    'OrderID': [1, 2, 2, 4, 5],
    'Region': ['South', 'West', 'West', None, 'South '],
    'Sales': [100.0, 250.0, 250.0, np.nan, 80.0],
    'OrderDate': ['2026-01-01', '2026-01-02', '2026-01-02', 'bad-date', '2026-01-05']
})
# Evaluate this expression output
print(df)
```

## 2.1 Row count analysis

**Concept:** Check record and field counts first.



```python
# 2.1 Row Count Analysis
# Definition: Measure record volume and overall dataset shape before deeper analysis.
# Example: Count total rows and view the (rows, columns) shape pair.

# Calculate total rows in the dataset
print('Rows:', len(df))
# Read dataset dimensions (rows, columns)
print('Shape (rows, columns):', df.shape)
```

## 2.2 Column / schema analysis

**Concept:** Inspect names, non-null counts, dtypes, and memory footprint.



```python
# 2.2 Column / Schema Analysis
# Definition: Inspect column names, null counts, data types, and memory usage.
# Example: List column names and review df.info() output for structure checks.

# Extract all column names
print(df.columns.tolist())
# Run this transformation step
df.info()
```

## 2.3 Data type analysis

**Concept:** Validate inferred types and decide whether parsing is required.



```python
# 2.3 Data Type Analysis
# Definition: Verify inferred data types and identify fields requiring conversion.
# Example: Run the cell and interpret the output against the section objective.

# Review inferred data types
print(df.dtypes)
```

## 2.4 NULL analysis

**Concept:** Count missing values and compare percentages across columns.



```python
# 2.4 NULL Analysis
# Definition: Quantify missing values to prioritize data-quality fixes.
# Example: Report both null counts and null percentages by column.

# Count missing values by column
print(df.isna().sum())
# Compute missing-value percentage by column
print((df.isna().mean() * 100).round(1))
```

## 2.5 Duplicate analysis

**Concept:** Measure full-row duplicates and key-level duplicates separately.



```python
# 2.5 Duplicate Analysis
# Definition: Detect repeated rows or repeated business keys that can bias analysis.
# Example: Compare full-row duplicates with key-level duplicates.

# Count duplicate rows or duplicate keys
print('Full duplicates:', df.duplicated().sum())
# Count duplicate rows or duplicate keys
print('Duplicate OrderIDs:', df['OrderID'].duplicated().sum())
```

## 2.6 Distinct / cardinality analysis

**Concept:** Count unique values to understand categorical diversity.



```python
# 2.6 Distinct / Cardinality Analysis
# Definition: Measure uniqueness and category concentration for categorical fields.
# Example: Run the cell and interpret the output against the section objective.

# Measure distinct values
print(df.nunique(dropna=True))
# Compute frequency distribution
print(df['Region'].value_counts(dropna=False))
```

## 2.7 Min / max analysis

**Concept:** Inspect numeric ranges to detect scale and boundary issues.



```python
# 2.7 Min / Max Analysis
# Definition: Check boundary values to detect implausible or extreme observations.
# Example: Run the cell and interpret the output against the section objective.

# Compute range using minimum and maximum values
print(df['Sales'].min(), df['Sales'].max())
```

## 2.8 Average / median analysis

**Concept:** Compare central tendency metrics; mean is sensitive to extremes.



```python
# 2.8 Average / Median Analysis
# Definition: Compare central tendency metrics and assess potential skew.
# Example: Compute mean and median Sales values, then compare their gap.

# Compute mean to assess central tendency
print('Mean:', df['Sales'].mean())
# Compute median for robust central tendency
print('Median:', df['Sales'].median())
```

## 2.9 Pattern / format analysis

**Concept:** Inspect text patterns and parseability; regex helps detect format issues.



```python
# 2.9 Pattern / Format Analysis
# Definition: Validate text/date format consistency and parseability.
# Example: Normalize text spacing and count invalid date parses.

# Convert data type for consistent analysis
print(df['Region'].astype('string').str.strip().value_counts(dropna=False))
# Parse text values into datetime and mark invalid entries as NaT
parsed = pd.to_datetime(df['OrderDate'], errors='coerce')
# Count missing values by column
print('Unparseable dates:', parsed.isna().sum())
```
