## Practice Dataset Setup

Run this setup cell first. It creates a reproducible sample dataset used by the techniques below.


```python
# 8.0 Practice Dataset Setup
# Definition: Create a reproducible sample dataset used across all examples in this notebook.
# Example: Build the sample DataFrame and confirm it loads before running section techniques.

# Import pandas
import pandas as pd
# Import numpy
import numpy as np
# Import matplotlib.pyplot
import matplotlib.pyplot as plt
# Import seaborn
import seaborn as sns
# Create the sample DataFrame
sales = pd.DataFrame({
 'Month': ['Jan','Jan','Jan','Feb','Feb','Feb','Mar','Mar','Mar'],
 'Region': ['North','South','West'] * 3,
 'Revenue': [1200, 1500, 900, 1300, 1700, 850, 1600, 1650, 1100],
 'Target': [1100, 1400, 1000, 1400, 1600, 1000, 1500, 1750, 1050],
 'Orders': [12, 15, 9, 13, 17, 8, 16, 16, 11]
})
# Run this transformation step
sales['Variance'] = sales['Revenue'] - sales['Target']
```

## 8.1 Descriptive analysis

**Concept:** Summarize level, spread and shape before interpreting.



```python
# 8.1 Descriptive Analysis
# Definition: Perform descriptive analysis to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Evaluate this expression output
print(sales['Revenue'].describe())
# Evaluate this expression output
print(sales.groupby('Region')['Revenue'].agg(['count','mean','median','sum']))
```

## 8.2 KPI analysis

**Concept:** Define metrics and denominators explicitly.



```python
# 8.2 KPI Analysis
# Definition: Analyze business metrics and explain performance changes.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
sales['AverageOrderValue'] = sales['Revenue'] / sales['Orders']
# Evaluate this expression output
print(sales[['Month','Region','Revenue','Orders','AverageOrderValue']])
```

## 8.3 Trend analysis

**Concept:** Aggregate by time in the correct order and visualize changes.



```python
# 8.3 Trend Analysis
# Definition: Explore distributions, patterns, and relationships before conclusions.
# Example: Run the cell and interpret the output against the section objective.

# Group records by key before summarization
monthly = sales.groupby('Month', sort=False)['Revenue'].sum()
# Run this transformation step
monthly.plot(marker='o', title='Monthly revenue trend')
# Run this transformation step
plt.ylabel('Revenue')
# Run this transformation step
plt.show()
```

## 8.4 Comparative analysis

**Concept:** Compare segments using the same metric and period.



```python
# 8.4 Comparative Analysis
# Definition: Perform comparative analysis to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Group records by key before summarization
comparison = sales.groupby('Region')['Revenue'].sum().sort_values(ascending=False)
# Run this transformation step
comparison.plot(kind='bar', title='Revenue by region')
# Run this transformation step
plt.show()
```

## 8.5 Contribution analysis

**Concept:** Calculate each segment's share of a defined total.



```python
# 8.5 Contribution Analysis
# Definition: Analyze business metrics and explain performance changes.
# Example: Run the cell and interpret the output against the section objective.

# Group records by key before summarization
region = sales.groupby('Region')['Revenue'].sum()
# Compute share
share = (region / region.sum() * 100).round(1)
# Evaluate this expression output
print(share)
```

## 8.6 Variance analysis

**Concept:** Compare actuals with targets or prior periods; define variance direction.



```python
# 8.6 Variance Analysis
# Definition: Analyze business metrics and explain performance changes.
# Example: Run the cell and interpret the output against the section objective.

# Group records by key before summarization
summary = sales.groupby('Region')[['Revenue','Target','Variance']].sum()
# Run this transformation step
summary['VariancePct'] = summary['Variance'] / summary['Target'] * 100
# Evaluate this expression output
print(summary.round(2))
```

## 8.7 Root cause analysis

**Concept:** Segment a problem, generate hypotheses and test them against evidence; do not infer causation from correlation alone.



```python
# 8.7 Root Cause Analysis
# Definition: Analyze business metrics and explain performance changes.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
sales['BelowTarget'] = sales['Revenue'] < sales['Target']
# Evaluate this expression output
print(sales.groupby(['Region','BelowTarget'])['Orders'].agg(['count','mean']))
# Run this transformation step
sns.scatterplot(data=sales, x='Orders', y='Revenue', hue='Region')
# Run this transformation step
plt.show()
```

## 8.8 Anomaly analysis

**Concept:** Compare observations with baselines or explicit thresholds and investigate exceptions.



```python
# 8.8 Anomaly Analysis
# Definition: Apply a data-mining method and evaluate model behavior.
# Example: Run the cell and interpret the output against the section objective.

# Run this transformation step
sales['RevenueZ'] = (sales['Revenue'] - sales['Revenue'].mean()) / sales['Revenue'].std(ddof=0)
# Evaluate this expression output
print(sales.loc[sales['RevenueZ'].abs() > 1.5, ['Month','Region','Revenue','RevenueZ']])
```
