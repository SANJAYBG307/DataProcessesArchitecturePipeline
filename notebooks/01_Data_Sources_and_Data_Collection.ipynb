# 1) Data Sources & Data Collection

This notebook is structured as one technique per cell.
Each technique cell includes: Definition, Example, and Practical Code.


```python
# Load pandas
import pandas as pd

# Load SQL engine helper
from sqlalchemy import create_engine

# Load request helper
import requests

# Load SQLite helper
import sqlite3

# Load in-memory text helper
from io import StringIO

# Load JSON helper
import json

# Note: Install optional connectors as needed
print("Install extras when needed: pymongo, redis, cassandra-driver, pandas-gbq, pyarrow, openpyxl")
```

## File-Based Data Sources


```python
# 1.1 CSV
# Definition: Collect tabular data from comma-separated files using pandas.
# Example: Load sales records from a CSV export.

# Set CSV path
csv_path = "path/to/file.csv"

# Read CSV with pandas
df_csv = pd.read_csv(csv_path)

# Show sample
print(df_csv.head())
```

```python
# 1.2 Excel
# Definition: Collect worksheet data from .xlsx files using pandas.
# Example: Load the Orders sheet from an Excel workbook.

# Set Excel path
excel_path = "path/to/file.xlsx"

# Set sheet name
sheet_name = "Sheet1"

# Read Excel with pandas
df_excel = pd.read_excel(excel_path, sheet_name=sheet_name)

# Show sample
print(df_excel.head())
```

```python
# 1.3 JSON
# Definition: Collect structured JSON data into a pandas DataFrame.
# Example: Load API response data saved as JSON.

# Set JSON path
json_path = "path/to/file.json"

# Read JSON with pandas
df_json = pd.read_json(json_path)

# Show sample
print(df_json.head())
```

```python
# 1.4 Parquet
# Definition: Collect columnar Parquet data for analytics workloads.
# Example: Load transaction data stored in Parquet format.

# Set Parquet path
parquet_path = "path/to/file.parquet"

# Read Parquet with pandas
df_parquet = pd.read_parquet(parquet_path)

# Show sample
print(df_parquet.head())
```

```python
# 1.5 XML Files
# Definition: Collect XML records by parsing nodes with an XPath.
# Example: Load order nodes from an XML document.

# Set XML path
xml_path = "path/to/file.xml"

# Set row xpath
row_xpath = ".//record"

# Read XML with pandas
df_xml = pd.read_xml(xml_path, xpath=row_xpath)

# Show sample
print(df_xml.head())
```

## RDBMS Data Sources


```python
# 1.6 MySQL
# Definition: Collect relational data from MySQL using SQL queries.
# Example: Run SELECT query and load result into pandas.

# Set MySQL connection string
mysql_conn = "mysql+pymysql://user:password@host:3306/database"

# Create SQL engine
mysql_engine = create_engine(mysql_conn)

# Set SQL query
mysql_query = "SELECT * FROM your_table"

# Read SQL result with pandas
df_mysql = pd.read_sql(mysql_query, mysql_engine)

# Show sample
print(df_mysql.head())
```

```python
# 1.7 PostgreSQL
# Definition: Collect relational data from PostgreSQL using SQL queries.
# Example: Read customer table from a PostgreSQL database.

# Set PostgreSQL connection string
postgres_conn = "postgresql+psycopg2://user:password@host:5432/database"

# Create SQL engine
postgres_engine = create_engine(postgres_conn)

# Set SQL query
postgres_query = "SELECT * FROM your_table"

# Read SQL result with pandas
df_postgres = pd.read_sql(postgres_query, postgres_engine)

# Show sample
print(df_postgres.head())
```

```python
# 1.8 SQL Server
# Definition: Collect relational data from SQL Server with ODBC.
# Example: Query top rows from a SQL Server table.

# Set SQL Server ODBC string
odbc_str = "DRIVER={ODBC Driver 17 for SQL Server};SERVER=host;DATABASE=db;UID=user;PWD=password"

# Build SQLAlchemy URL
sqlserver_conn = "mssql+pyodbc:///?odbc_connect=" + requests.utils.quote(odbc_str)

# Create SQL engine
sqlserver_engine = create_engine(sqlserver_conn)

# Set SQL query
sqlserver_query = "SELECT TOP 100 * FROM your_table"

# Read SQL result with pandas
df_sqlserver = pd.read_sql(sqlserver_query, sqlserver_engine)

# Show sample
print(df_sqlserver.head())
```

```python
# 1.9 SQLite
# Definition: Collect lightweight relational data from SQLite databases.
# Example: Read local application data from a .db file.

# Set SQLite database path
sqlite_path = "path/to/file.db"

# Open SQLite connection
sqlite_conn = sqlite3.connect(sqlite_path)

# Set SQL query
sqlite_query = "SELECT * FROM your_table"

# Read SQL result with pandas
df_sqlite = pd.read_sql_query(sqlite_query, sqlite_conn)

# Close connection
sqlite_conn.close()

# Show sample
print(df_sqlite.head())
```

```python
# 1.10 Oracle
# Definition: Collect enterprise relational data from Oracle databases.
# Example: Run SQL query on Oracle and load to pandas.

# Set Oracle connection string
oracle_conn = "oracle+oracledb://user:password@host:1521/?service_name=service"

# Create SQL engine
oracle_engine = create_engine(oracle_conn)

# Set SQL query
oracle_query = "SELECT * FROM your_table FETCH FIRST 100 ROWS ONLY"

# Read SQL result with pandas
df_oracle = pd.read_sql(oracle_query, oracle_engine)

# Show sample
print(df_oracle.head())
```

## NoSQL Data Sources


```python
# 1.11 MongoDB
# Definition: Collect document data from MongoDB collections.
# Example: Fetch JSON-like documents and normalize them.

# Load MongoDB client
from pymongo import MongoClient

# Set MongoDB connection
mongo_uri = "mongodb://user:password@host:27017"

# Connect MongoDB
mongo_client = MongoClient(mongo_uri)

# Select collection
mongo_collection = mongo_client["database"]["collection"]

# Read documents
mongo_docs = list(mongo_collection.find({}, {"_id": 0}))

# Convert docs to DataFrame
df_mongodb = pd.json_normalize(mongo_docs)

# Show sample
print(df_mongodb.head())
```

```python
# 1.12 Cassandra
# Definition: Collect wide-column data from Cassandra keyspaces.
# Example: Read rows from a Cassandra table into pandas.

# Load Cassandra cluster helper
from cassandra.cluster import Cluster

# Connect Cassandra cluster
cassandra_cluster = Cluster(["host1", "host2"])

# Open Cassandra session
cassandra_session = cassandra_cluster.connect("your_keyspace")

# Run CQL query
cassandra_rows = cassandra_session.execute("SELECT * FROM your_table")

# Convert rows to DataFrame
df_cassandra = pd.DataFrame(list(cassandra_rows))

# Show sample
print(df_cassandra.head())
```

```python
# 1.13 Redis
# Definition: Collect key-value records from Redis data structures.
# Example: Load hash records by key prefix.

# Load Redis client
import redis

# Connect Redis
redis_client = redis.Redis(host="host", port=6379, password="password", decode_responses=True)

# Get keys
redis_keys = redis_client.keys("prefix:*")

# Read hash records
redis_records = [redis_client.hgetall(k) for k in redis_keys]

# Convert records to DataFrame
df_redis = pd.DataFrame(redis_records)

# Show sample
print(df_redis.head())
```

## Cloud Storage / Data Lakes


```python
# 1.14 Amazon S3
# Definition: Collect files from Amazon S3 object storage.
# Example: Read CSV directly from an S3 URI.

# Set S3 file path
s3_path = "s3://bucket/path/file.csv"

# Read S3 CSV with pandas
df_s3 = pd.read_csv(
    s3_path,
    storage_options={"key": "AWS_ACCESS_KEY", "secret": "AWS_SECRET_KEY"}
)

# Show sample
print(df_s3.head())
```

```python
# 1.15 Azure Data Lake Storage
# Definition: Collect files from Azure Data Lake Storage paths.
# Example: Read Parquet from an ADLS container path.

# Set ADLS file path
adls_path = "abfss://container@account.dfs.core.windows.net/folder/file.parquet"

# Read ADLS parquet with pandas
df_adls = pd.read_parquet(
    adls_path,
    storage_options={"account_name": "account", "account_key": "account_key"}
)

# Show sample
print(df_adls.head())
```

```python
# 1.16 Google Cloud Storage
# Definition: Collect files from Google Cloud Storage buckets.
# Example: Read CSV from a gs:// object path.

# Set GCS file path
gcs_path = "gs://bucket/path/file.csv"

# Read GCS CSV with pandas
df_gcs = pd.read_csv(
    gcs_path,
    storage_options={"token": "path/to/gcp-service-account.json"}
)

# Show sample
print(df_gcs.head())
```

## API Data Sources


```python
# 1.21 REST APIs
# Definition: Collect API data from REST endpoints over HTTP.
# Example: GET JSON response and normalize data field.

# Set REST endpoint
rest_url = "https://api.example.com/v1/items"

# Set API headers
rest_headers = {"Authorization": "Bearer YOUR_TOKEN"}

# Call REST API
rest_response = requests.get(rest_url, headers=rest_headers, timeout=30)

# Raise error for bad status
rest_response.raise_for_status()

# Parse JSON response
rest_payload = rest_response.json()

# Convert JSON list to DataFrame
df_rest = pd.json_normalize(rest_payload["data"])

# Show sample
print(df_rest.head())
```

```python
# 1.22 SOAP APIs
# Definition: Collect API data from SOAP services using XML envelopes.
# Example: POST SOAP request and parse XML response.

# Set SOAP endpoint
soap_url = "https://service.example.com/soap"

# Set SOAP header
soap_headers = {"Content-Type": "text/xml; charset=utf-8"}

# Build SOAP request XML
soap_body = """<soap:Envelope xmlns:soap='http://schemas.xmlsoap.org/soap/envelope/'><soap:Body><GetItems/></soap:Body></soap:Envelope>"""

# Call SOAP API
soap_response = requests.post(soap_url, data=soap_body, headers=soap_headers, timeout=30)

# Raise error for bad status
soap_response.raise_for_status()

# Read XML response into DataFrame
df_soap = pd.read_xml(StringIO(soap_response.text), xpath=".//item")

# Show sample
print(df_soap.head())
```
