## Practice Dataset Setup

Run this setup cell first. It creates a reproducible sample dataset used by the techniques below.


```python
# 5.0 Practice Dataset Setup
# Definition: Create a reproducible sample dataset used across all examples in this notebook.
# Example: Build the sample DataFrame and confirm it loads before running section techniques.

# Import pandas
import pandas as pd
# Create the sample DataFrame
orders = pd.DataFrame({
    'OrderID': [1, 2, 3, 4],
    'Region': ['South', 'West', 'South', 'North'],
    'Product': ['Laptop', 'Mouse', 'Laptop', 'Keyboard'],
    'Quantity': [1, 2, 2, 3],
    'UnitPrice': [750, 25, 700, 45],
    'OrderDate': pd.to_datetime(['2026-01-03', '2026-01-04', '2026-02-02', '2026-02-10'])
})
# Run this transformation step
orders['Sales'] = orders['Quantity'] * orders['UnitPrice']
# Create the sample DataFrame
customers = pd.DataFrame({'Region': ['South', 'West', 'North'], 'Manager': ['Kim', 'Lee', 'Sam']})
```

## 5.1 Select columns

**Concept:** Keep only the fields needed for a task.



```python
# 5.1 Select Columns
# Definition: Inspect column names, null counts, data types, and memory usage.
# Example: List column names and review df.info() output for structure checks.

# Evaluate this expression output
print(orders[['OrderID', 'Region', 'Sales']])
```

## 5.2 Filter rows

**Concept:** Use boolean conditions to select records.



```python
# 5.2 Filter Rows
# Definition: Perform filter rows to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Evaluate this expression output
print(orders.loc[orders['Sales'] >= 100, ['OrderID', 'Sales']])
```

## 5.3 Create derived columns

**Concept:** Calculate fields from existing values.



```python
# 5.3 Create Derived Columns
# Definition: Inspect column names, null counts, data types, and memory usage.
# Example: List column names and review df.info() output for structure checks.

# Run this transformation step
orders['HighValue'] = orders['Sales'] >= 500
# Evaluate this expression output
print(orders[['OrderID', 'Sales', 'HighValue']])
```

## 5.4 Group By

**Concept:** Group records by category or time period.



```python
# 5.4 Group By
# Definition: Reshape data into analysis-ready structures while preserving correctness.
# Example: Run the cell and interpret the output against the section objective.

# Evaluate this expression output
print(orders.groupby('Region')['Sales'].sum())
```

## 5.5 Aggregation

**Concept:** Compute multiple summary metrics.



```python
# 5.5 Aggregation
# Definition: Reshape data into analysis-ready structures while preserving correctness.
# Example: Run the cell and interpret the output against the section objective.

# Evaluate this expression output
print(orders.groupby('Region').agg(orders=('OrderID', 'count'), revenue=('Sales', 'sum'), avg_order=('Sales', 'mean')) )
```

## 5.6 Join

**Concept:** Combine tables on a shared key and validate cardinality when possible.



```python
# 5.6 Join
# Definition: Reshape data into analysis-ready structures while preserving correctness.
# Example: Merge transactional data with a lookup table using join keys.

# Join two datasets on key columns
joined = orders.merge(customers, on='Region', how='left', validate='many_to_one')
# Evaluate this expression output
print(joined[['OrderID', 'Region', 'Manager']])
```

## 5.7 Pivot

**Concept:** Reshape categories into columns; `pivot_table` aggregates duplicate combinations.



```python
# 5.7 Pivot
# Definition: Reshape data into analysis-ready structures while preserving correctness.
# Example: Pivot grouped values into a matrix-style summary table.

# Evaluate this expression output
print(orders.pivot_table(index='Region', columns='Product', values='Sales', aggfunc='sum', fill_value=0))
```

## 5.8 Unpivot

**Concept:** Convert wide columns into long rows with `melt()`.



```python
# 5.8 Unpivot
# Definition: Reshape data into analysis-ready structures while preserving correctness.
# Example: Pivot grouped values into a matrix-style summary table.

# Create the sample DataFrame
wide = pd.DataFrame({'Region': ['South', 'West'], 'Jan': [100, 80], 'Feb': [120, 90]})
# Unpivot wide columns into long format
long = wide.melt(id_vars='Region', var_name='Month', value_name='Sales')
# Evaluate this expression output
print(long)
```

## 5.9 Window functions

**Concept:** Use rolling, expanding, ranking, shift and grouped transforms for ordered calculations.



```python
# 5.9 Window Functions
# Definition: Perform window functions to understand and validate the dataset.
# Example: Run the cell and interpret the output against the section objective.

# Compute orders
orders = orders.sort_values('OrderDate')
# Run this transformation step
orders['CumulativeSales'] = orders['Sales'].cumsum()
# Run this transformation step
orders['PreviousOrderSales'] = orders['Sales'].shift(1)
# Evaluate this expression output
print(orders[['OrderDate', 'Sales', 'CumulativeSales', 'PreviousOrderSales']])
```
