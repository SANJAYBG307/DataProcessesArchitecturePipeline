## Practice Dataset Setup

Run this setup cell first. It creates a reproducible sample dataset used by the techniques below.


```python
# 9.0 Practice Dataset Setup
# Definition: Create a reproducible sample dataset used across all examples in this notebook.
# Example: Build the sample DataFrame and confirm it loads before running section techniques.

# Import pandas
import pandas as pd
# Import numpy
import numpy as np
# Import required helpers from sklearn.model_selection
from sklearn.model_selection import train_test_split
# Import required helpers from sklearn.pipeline
from sklearn.pipeline import make_pipeline
# Import required helpers from sklearn.preprocessing
from sklearn.preprocessing import StandardScaler
# Import required helpers from sklearn.impute
from sklearn.impute import SimpleImputer
# Import required helpers from sklearn.linear_model
from sklearn.linear_model import LogisticRegression, LinearRegression
# Import required helpers from sklearn.cluster
from sklearn.cluster import KMeans
# Import required helpers from sklearn.ensemble
from sklearn.ensemble import IsolationForest
# Import required helpers from sklearn.metrics
from sklearn.metrics import accuracy_score, classification_report, mean_absolute_error, r2_score, silhouette_score
# Compute rng
rng = np.random.default_rng(10)
# Compute n
n = 180
# Create the sample DataFrame
df = pd.DataFrame({
 'Quantity': rng.integers(1, 8, n),
 'UnitPrice': rng.choice([20, 40, 100, 500, 800], n).astype(float),
 'Discount': rng.uniform(0, .2, n),
 'RegionCode': rng.integers(0, 3, n)
})
# Run this transformation step
df['Sales'] = df['Quantity'] * df['UnitPrice'] * (1 - df['Discount'])
# Convert data type for consistent analysis
df['HighSales'] = (df['Sales'] >= 300).astype(int)
# Compute features
features = ['Quantity', 'UnitPrice', 'Discount', 'RegionCode']
```

## 9.1 Classification

**Concept:** Predict a category; evaluate on held-out data using suitable metrics.



```python
# 9.1 Classification
# Definition: Apply a data-mining method and evaluate model behavior.
# Example: Run the cell and interpret the output against the section objective.

# Split data into training and testing subsets
X_train, X_test, y_train, y_test = train_test_split(df[features], df['HighSales'], test_size=0.25, random_state=42, stratify=df['HighSales'])
# Compute model
model = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), LogisticRegression(max_iter=1000))
# Fit model/transformer parameters
model.fit(X_train, y_train)
# Generate predictions from the model
pred = model.predict(X_test)
# Evaluate this expression output
print('Accuracy:', round(accuracy_score(y_test, pred), 3))
# Evaluate this expression output
print(classification_report(y_test, pred, zero_division=0))
```

## 9.2 Clustering

**Concept:** Group observations without known labels; scale features when distances are scale-sensitive.



```python
# 9.2 Clustering
# Definition: Apply a data-mining method and evaluate model behavior.
# Example: Run the cell and interpret the output against the section objective.

# Compute X
X = df[['Quantity','UnitPrice','Discount','Sales']]
# Fit transformer/model and transform data in one step
X_scaled = StandardScaler().fit_transform(X)
# Compute km
km = KMeans(n_clusters=3, random_state=42, n_init=10)
# Generate predictions from the model
labels = km.fit_predict(X_scaled)
# Compute frequency distribution
print(pd.Series(labels).value_counts().sort_index())
# Evaluate this expression output
print('Silhouette:', round(silhouette_score(X_scaled, labels), 3))
```

## 9.3 Regression

**Concept:** Predict a continuous numeric outcome and evaluate on held-out data.



```python
# 9.3 Regression
# Definition: Apply a data-mining method and evaluate model behavior.
# Example: Run the cell and interpret the output against the section objective.

# Split data into training and testing subsets
X_train, X_test, y_train, y_test = train_test_split(df[features], df['Sales'], test_size=0.25, random_state=42)
# Compute model
model = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), LinearRegression())
# Fit model/transformer parameters
model.fit(X_train, y_train)
# Generate predictions from the model
pred = model.predict(X_test)
# Evaluate this expression output
print('MAE:', round(mean_absolute_error(y_test, pred), 2))
# Evaluate this expression output
print('R2:', round(r2_score(y_test, pred), 3))
```

## 9.4 Anomaly / outlier detection

**Concept:** Flag unusual records; review flagged observations and tune contamination based on context.



```python
# 9.4 Anomaly / Outlier Detection
# Definition: Apply a data-mining method and evaluate model behavior.
# Example: Run the cell and interpret the output against the section objective.

# Compute X
X = df[['Quantity','UnitPrice','Discount','Sales']]
# Fit transformer/model and transform data in one step
X_scaled = StandardScaler().fit_transform(X)
# Compute detector
detector = IsolationForest(contamination=0.03, random_state=42)
# Generate predictions from the model
df['AnomalyFlag'] = detector.fit_predict(X_scaled) == -1
# Evaluate this expression output
print(df.loc[df['AnomalyFlag'], ['Quantity','UnitPrice','Discount','Sales']].head())
# Evaluate this expression output
print('Flagged:', int(df['AnomalyFlag'].sum()))
```
